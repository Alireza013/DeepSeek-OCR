# **Requirements**

```python
from pathlib import Path

requirements_file = next(Path.cwd().rglob("requirements.txt"), None)
if requirements_file is None:
    raise FileNotFoundError(
        f"requirements.txt was not found below the current directory: {Path.cwd()}"
    )

print(f"Installing from: {requirements_file.resolve()}")
%pip install -r {requirements_file}
```
